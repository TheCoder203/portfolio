## JSON Parsing

Learning from AstroPlatformer

```javascript
%%js
const obj = {name:"Level1", width:650};
const json = JSON.stringify(obj);
console.log("JSON:", json);
const parsed = JSON.parse(json);
console.log("Parsed:", parsed.name);
```
