## Methods and Parameters

Learning from AstroPlatformer

```javascript
%%js
function scale(obj, sx, sy) {
  return {id:obj.id, x:obj.x*sx, y:obj.y*sy};
}
const result = scale({id:"p1",x:40,y:290}, 1.5, 1.5);
console.log(result);
```
