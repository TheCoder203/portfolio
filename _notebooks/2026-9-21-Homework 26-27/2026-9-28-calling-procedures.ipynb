```python
# CODE_RUNNER: popcorn hack 2# Helper Proceduredef is_priority(incident_type):    # Return True when the incident is a "Medical Emergency"    return incident_type == "Medical Emergency"# Outer Proceduredef process_report(location, incident_type):    # Call is_priority() using incident_type and store the returned value    is_crit = is_priority(incident_type)        # Use an if/else statement    if is_crit:        print(f"URGENT: High-priority incident at {location} — Medical Emergency!")    else:        print(f"Standard Report: Incident at {location} logged as '{incident_type}'.")# Call process_report() two times with different incident typesprocess_report("Sector 7", "Medical Emergency")process_report("Sector 4", "Maintenance Issue")
```
