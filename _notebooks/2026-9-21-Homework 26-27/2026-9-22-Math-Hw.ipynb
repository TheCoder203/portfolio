```python
# CODERUNNER: homeowrk


# Recovery Center Cost Report

cost_per_session = 35
sessions_per_week = 3
weeks_per_month = 4
enrollment_fee = 75

# 1. Calculate weekly cost
weekly = cost_per_session * sessions_per_week

# 2. Calculate monthly cost
monthly = weekly * weeks_per_month

# 3. Calculate total cost with fee
fee = monthly + enrollment_fee

# 4. Calculate average daily cost
daily = monthly / 30

# Print all results with labels
print("Weekly Cost: ", weekly)
print("Monthly Cost: ", monthly)
print("Total Cost with Fee: ", fee)
print("Average Daily Cost: ", daily)
```
