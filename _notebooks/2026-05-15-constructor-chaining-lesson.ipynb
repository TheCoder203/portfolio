## Constructor Chaining

Learning from AstroPlatformer

```javascript
%%js
class Coin {
  constructor(id, x) {
    this.id = id; this.x = x;
  }
}
class GameCoin extends Coin {
  constructor(id, x, value) {
    super(id, x);
    this.value = value;
  }
}
const c = new GameCoin("c1", 90, 1);
console.log(c);
```
